```python
import pandas as pd
from pypinyin import lazy_pinyin

def convert_to_pinyin(text):
    """
    将文本中的中文字符转换为首字母大写的拼音，保留非中文字符。
    """
    # 处理空值 (NaN)
    if pd.isna(text):
        return text
    
    text = str(text)
    result = []
    
    for char in text:
        # 判断是否为中文字符 (Unicode 范围)
        if '\u4e00' <= char <= '\u9fff':
            # 获取拼音并将其首字母大写
            pinyin_str = lazy_pinyin(char)[0]
            result.append(pinyin_str.capitalize())
        else:
            # 非中文字符（如数字、英文字母、括号等）保持原样
            result.append(char)
            
    return "".join(result)

def main():
    # 1. 读取你的 Excel 文件
    file_path = 'Processed Data.xlsx'
    print(f"正在读取文件: {file_path}...")
    df = pd.read_excel(file_path)
    
    # 2. 指定需要转换的列名
    # 根据你的截图，这四列包含需要转换的中文
    columns_to_convert = [
        "Province(SAP Field REGION)", 
        "City(SAP Field CITY1)", 
        "District(SAP Field CITY2)", 
        "Address(SAP Field STREET)"
    ]
    
    # 3. 对指定列应用转换函数
    for col in columns_to_convert:
        if col in df.columns:
            df[col] = df[col].apply(convert_to_pinyin)
        else:
            print(f"警告: 未在数据中找到列名 '{col}'，已跳过。")
            
    # 4. 保存为新的 Excel 文件
    output_path = 'Pinyin_Processed_Data.xlsx'
    df.to_excel(output_path, index=False, engine='openpyxl')
    print(f"转换完成！文件已保存为: {output_path}")

if __name__ == "__main__":
    main()
```
